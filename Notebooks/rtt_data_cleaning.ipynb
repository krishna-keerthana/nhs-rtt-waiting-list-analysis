```python
import re

def process_month(zip_path):
    # Get the month label from the zip filename
    match = re.search(r'rtt_(\d{4})_(\d{2})', zip_path)
    year, month = match.group(1), match.group(2)
    month_label = f"{year}-{month}"

    # Extract the zip into temp folder
    temp = '/content/temp_unzip'
    os.makedirs(temp, exist_ok=True)
    with zipfile.ZipFile(zip_path, 'r') as z:
        z.extractall(temp)

    # Find the CSV inside and load it
    csv_name = [f for f in os.listdir(temp) if f.endswith('.csv')][0]
    df = pd.read_csv(temp + '/' + csv_name)

    # Keep only Incomplete Pathways
    inc = df[df['RTT Part Type'] == 'Part_2'].copy()

    # Build the summary columns
    wait_cols = [c for c in inc.columns if 'Gt' in c and 'Weeks' in c]
    within_18 = [c for c in wait_cols if any(f'{i:02d} To' in c for i in range(0, 18))]
    over_52 = [c for c in wait_cols if any(f'{i} To' in c or f'Gt {i} Weeks' in c for i in range(52, 105))]

    inc['total_waiting'] = inc[wait_cols].sum(axis=1)
    inc['waiting_within_18wks'] = inc[within_18].sum(axis=1)
    inc['waiting_over_52wks'] = inc[over_52].sum(axis=1)
    inc['month'] = month_label

    # Clean up temp folder for the next month
    for f in os.listdir(temp):
        os.remove(temp + '/' + f)

    keep = ['month', 'Provider Org Code', 'Provider Org Name',
            'Treatment Function Code', 'Treatment Function Name',
            'total_waiting', 'waiting_within_18wks', 'waiting_over_52wks']
    return inc[keep]


# Run the function for every zip in the folder
all_months = []
for f in sorted(os.listdir(raw_data_path)):
    print(f"Processing {f} ...")
    month_data = process_month(raw_data_path + '/' + f)
    all_months.append(month_data)
    print(f"   Rows: {len(month_data)}")

# Combine every month into one big clean file
combined = pd.concat(all_months, ignore_index=True)
combined.to_csv(clean_path + '/rtt_all_months_clean.csv', index=False)

print(f"\nAll done. Combined shape: {combined.shape}")
```

```python
# How many rows per month?
print(combined['month'].value_counts().sort_index())

# Any totally empty rows?
print("\nRows with zero patients:", (combined['total_waiting'] == 0).sum())

# How many unique trusts and specialties?
print("\nUnique trusts:", combined['Provider Org Name'].nunique())
print("Unique specialties:", combined['Treatment Function Name'].nunique())
```

```python
# Group by month, trust, specialty and sum the patient counts
agg = combined.groupby(['month', 'Provider Org Code', 'Provider Org Name',
                        'Treatment Function Code', 'Treatment Function Name'],
                       as_index=False).agg({
    'total_waiting': 'sum',
    'waiting_within_18wks': 'sum',
    'waiting_over_52wks': 'sum'
})

print("Aggregated shape:", agg.shape)
print(agg.head(10))
```

```python
# Separate the two types
specialty_level = agg[agg['Treatment Function Code'] != 'C_999'].copy()
trust_level = agg[agg['Treatment Function Code'] == 'C_999'].copy()

# Add a compliance % column — the key NHS metric
specialty_level['pct_within_18wks'] = (
    specialty_level['waiting_within_18wks'] / specialty_level['total_waiting'] * 100
).round(1)

trust_level['pct_within_18wks'] = (
    trust_level['waiting_within_18wks'] / trust_level['total_waiting'] * 100
).round(1)

# Save both to Drive
specialty_level.to_csv(clean_path + '/rtt_by_specialty.csv', index=False)
trust_level.to_csv(clean_path + '/rtt_by_trust.csv', index=False)

print("Specialty-level file:", specialty_level.shape)
print("Trust-level file:", trust_level.shape)
```

```python
# The total waiting list change over 12 months
monthly_total = specialty_level.groupby('month').agg({
    'total_waiting': 'sum',
    'waiting_within_18wks': 'sum',
    'waiting_over_52wks': 'sum'
}).reset_index()

monthly_total['pct_within_18wks'] = (
    monthly_total['waiting_within_18wks'] / monthly_total['total_waiting'] * 100
).round(1)

print(monthly_total)
```

```python
# specialties having the longest waiting lists AND worst 18-week compliance
latest = specialty_level[specialty_level['month'] == '2026-03']

# Sum across all trusts to get an England-wide view by specialty
by_spec = latest.groupby('Treatment Function Name').agg({
    'total_waiting': 'sum',
    'waiting_within_18wks': 'sum',
    'waiting_over_52wks': 'sum'
}).reset_index()

by_spec['pct_within_18wks'] = (
    by_spec['waiting_within_18wks'] / by_spec['total_waiting'] * 100
).round(1)

# Sort by worst compliance
by_spec_sorted = by_spec.sort_values('pct_within_18wks').head(10)
print(by_spec_sorted[['Treatment Function Name', 'total_waiting',
                       'waiting_over_52wks', 'pct_within_18wks']])
```

```python
# Trusts having the most patients waiting over 52 weeks
latest_trust = trust_level[trust_level['month'] == '2026-03']

worst_trusts = latest_trust.nlargest(10, 'waiting_over_52wks')
print(worst_trusts[['Provider Org Name', 'total_waiting',
                     'waiting_over_52wks', 'pct_within_18wks']])
```

```python
# For each specialty, see the monthly trend across all 12 months
specialty_trend = specialty_level.groupby(['month', 'Treatment Function Name']).agg({
    'total_waiting': 'sum',
    'waiting_within_18wks': 'sum',
    'waiting_over_52wks': 'sum'
}).reset_index()

specialty_trend['pct_within_18wks'] = (
    specialty_trend['waiting_within_18wks'] / specialty_trend['total_waiting'] * 100
).round(1)

specialty_trend.to_csv(clean_path + '/rtt_specialty_trend.csv', index=False)
print("Specialty trend file:", specialty_trend.shape)
print(specialty_trend.head(10))
```
